## **Importing Libraries**

```python
import warnings
warnings.filterwarnings("ignore")
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option("display.max_columns", 20)
```

## *Features*

| Column | Type | Description |
|--------|------|-------------|
| Major_Category | Categorical | Student's major (STEM, Business, Humanities, Medical, Arts) |
| Year_of_Study | Categorical | Academic year |
| Pre_Semester_GPA | Float | GPA before the semester |
| Post_Semester_GPA | Float | **Target:** GPA after the semester |
| Weekly_GenAI_Hours | Float | Weekly hours spent using Generative AI |
| Primary_Use_Case | Categorical | Main purpose of AI usage |
| Prompt_Engineering_Skill | Categorical | Prompt writing proficiency |
| Tool_Diversity | Integer | Number of AI tools used |
| Paid_Subscription | Boolean | Paid AI subscription (Yes/No) |
| Traditional_Study_Hours | Float | Weekly study hours without AI |
| Perceived_AI_Dependency | Integer | AI dependency score (1–10) |
| Institutional_Policy | Categorical | Institution's AI policy |
| Anxiety_Level_During_Exams | Integer | Exam anxiety score (1–10) |
| Skill_Retention_Score | Float | Skill retention score (0–100) |
| Burnout_Risk_Level | Categorical | Burnout risk (Low, Medium, High) |

## **Exploratory Data Analysis**

### *Univariate Analysis*

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df,
    x="Weekly_GenAI_Hours",
    kde=True,
    bins=30,
    color="#4C72B0",
    edgecolor="black",
    alpha=0.7,
    line_kws={"linewidth": 2}
)

plt.title("Students' AI Usage Per Week", fontsize=18, fontweight="bold")
plt.xlabel("Weekly GenAI Hours", fontsize=12)
plt.ylabel("Count", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

ax = sns.countplot(
    data=df,
    x="Paid_Subscription",
    edgecolor="black",
    width=0.3,
    hue="Paid_Subscription",
    palette="mako",
    legend=False
)

total = len(df)

for p in ax.patches:
    percentage = 100 * p.get_height() / total
    ax.annotate(f'{percentage:.1f}%',
                (p.get_x() + p.get_width()/2, p.get_height()),
                ha='center',
                va='bottom',
                fontsize=11)

plt.title("Paid Subscription Distribution", fontsize=18, fontweight="bold")
plt.xlabel("Paid Subscription", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.5)

sns.despine()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.kdeplot(
    data=df,
    x="Traditional_Study_Hours",
    fill=True,
    linewidth=2,
    alpha=0.6,
    color="#4C72B0"
)

plt.title("Distribution of Traditional Study Hours", fontsize=18, fontweight="bold")
plt.xlabel("Traditional Study Hours per Week", fontsize=12)
plt.ylabel("Density", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

ax = sns.countplot(
    data=df,
    x="Primary_Use_Case",
    edgecolor="black",
    width=0.3,
    hue="Primary_Use_Case",
    palette="mako",
    legend=False
)
total = len(df)
for p in ax.patches:
    percentage = 100 * p.get_height() / total
    ax.annotate(f'{percentage:.1f}%',
                (p.get_x() + p.get_width() / 2, p.get_height()),
                ha='center',
                va='bottom',
                fontsize=11)

plt.title("Primary Use Case", fontsize=18, fontweight=600)
plt.grid(ls="--", alpha=0.5, axis="y")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Tool_Diversity",
    width=0.4,
    linewidth=1.5,
    color="#4C72B0"
)

plt.title("Distribution of AI Tool Diversity", fontsize=18, fontweight="bold")
plt.xlabel("Number of AI Tools Used", fontsize=12)

plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

ax = sns.countplot(
    data=df,
    x="Prompt_Engineering_Skill",
    edgecolor="black",
    width=0.35,
    palette="mako"
)

total = len(df)

for p in ax.patches:
    percentage = 100 * p.get_height() / total
    ax.annotate(
        f"{percentage:.1f}%",
        (p.get_x() + p.get_width()/2, p.get_height()),
        ha="center",
        va="bottom",
        fontsize=11
    )

plt.title("Distribution of Prompt Engineering Skill", fontsize=18, fontweight="bold")
plt.xlabel("Prompt Engineering Skill", fontsize=12)
plt.ylabel("Count", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="Perceived_AI_Dependency",
    color="#4C72B0",
    linewidth=1.5,
    inner="box"
)

plt.title("Distribution of Perceived AI Dependency", fontsize=18, fontweight="bold")
plt.xlabel("Perceived AI Dependency (1-10)", fontsize=12)

plt.grid(axis="x", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### *Bivariate Analysis*

```python
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x="Pre_Semester_GPA",
    y="Post_Semester_GPA",
    scatter_kws={"alpha": 0.4},
    line_kws={"color": "red"},
    color="#009E73"
)

plt.title("Pre Semester GPA vs Post Semester GPA", fontsize=18, fontweight="bold")
plt.grid(ls="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
